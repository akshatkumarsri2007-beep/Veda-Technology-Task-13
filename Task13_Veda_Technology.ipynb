```python
import pandas as pd

df = pd.read_csv("/content/retail_sales_dataset.csv")
print("Shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
print("Full-row duplicates:", df.duplicated().sum())
full_dup = df[df.duplicated(keep=False)].sort_values("Order_ID")
full_dup.head(10)
```

```python
print("Order_ID duplicates:", df.duplicated(subset=["Order_ID"]).sum())
key_dup = df[df.duplicated(subset=["Order_ID"], keep=False)].sort_values("Order_ID")
key_dup.head(10)
```

```python
with pd.ExcelWriter("duplicate_report.xlsx") as w:
    full_dup.to_excel(w, sheet_name="Full_Row_Duplicates", index=False)
    key_dup.to_excel(w, sheet_name="OrderID_Duplicates", index=False)
```

```python
clean = df.drop_duplicates()
clean = clean.drop_duplicates(subset=["Order_ID"], keep="first")
clean.to_csv("cleaned_retail_sales.csv", index=False)
print("Rows after cleaning:", len(clean))
```

```python
note = f"""AUDIT NOTE - Task 13: Duplicate Record Check
Original rows       : {len(df)}
Full-row duplicates : {df.duplicated().sum()}
Order_ID duplicates : {df.duplicated(subset=['Order_ID']).sum()}
Rows after cleaning : {len(clean)}
Method: drop_duplicates() on full rows, then on Order_ID (keep first).
Assumption: Order_ID should be unique per order.
"""
open("audit_note.txt", "w").write(note)
print(note)
```

**Conclusion**

In this task, I analyzed the Retail Sales dataset (1,060 records, 10 columns) to identify and document duplicate records using Python (Pandas).

**Two types of duplicates were found:**

Full-row duplicates (40 records): Every column value was identical to another record. These were exact copies, most likely caused by data entry or system errors, and were safely removed.
Key-column duplicates (20 records): The Order_ID was repeated, but other values such as City or Payment_Mode differed. These are conflicting records, since an Order_ID should be unique. I kept the first occurrence and documented the rest in the duplicate report so they can be verified against the source system.

After cleaning, the dataset had 1,000 unique records. This shows that duplicates can inflate sales totals and mislead analysis if not handled. Duplicate detection should be checked both on full rows and on key columns, because each method catches different problems.

**Learning outcome:** I learned how duplicated() and drop_duplicates() work, the role of the keep and subset parameters, and why documenting every cleaning step matters for a proper audit trail.
